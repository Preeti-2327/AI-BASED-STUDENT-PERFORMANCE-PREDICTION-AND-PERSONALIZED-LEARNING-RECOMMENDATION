# 01 — Exploratory Data Analysis

Canonical dataset + schema come from `ml-service/dataset.py` (`data/student-mat.csv`, UCI Student-Mat, target `G3`).

```python
import os, sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, os.path.abspath('..'))
from dataset import load_dataset

data, SCHEMA, DATA_PATH = load_dataset()
TARGET = SCHEMA['target']

print(f'Schema: {SCHEMA["name"]} | Target: {TARGET} (0-{SCHEMA["scale_max"]})')
print(f"Dataset: {DATA_PATH}")
print('Shape:', data.shape)
```

## Target distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=data[TARGET], bins=21, kde=True)
plt.title(f'Distribution of final grades ({TARGET})')
plt.xlabel(f'Target ({TARGET})')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## Missing values & correlations

```python
print('Missing values:\n', data.isnull().sum())

numeric = data.select_dtypes(include=['int64', 'float64'])
corr = numeric.corr()
plt.figure(figsize=(14, 10))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation heatmap')
plt.tight_layout()
plt.show()
```
